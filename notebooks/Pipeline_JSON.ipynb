# Базовый отчёт AquaBit: JSON + digWater

Этот ноутбук строит новый фактический отчёт без CSV и INP. Он не заменяет старую ИИ-диагностику и карту.


```python
from pathlib import Path
import sys

# Запускайте из корня проекта или из notebooks/.
_cwd = Path.cwd().resolve()
PROJECT_ROOT = _cwd.parent if _cwd.name.lower() == "notebooks" else _cwd
SRC_DIR = PROJECT_ROOT / "src"
if not (SRC_DIR / "reporting" / "json_report.py").is_file():
    raise FileNotFoundError(f"Не найден src/reporting/json_report.py в {PROJECT_ROOT}")
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

CITY_NAME = "модель семей"
CITY_DIR = PROJECT_ROOT / "data" / CITY_NAME
REPORT_JSON = CITY_DIR / "Report.json"
DIGWATER = CITY_DIR / "Модель Семей 32 для ИИ.digWater"
OUTPUT_HTML = PROJECT_ROOT / "reports" / f"AquaBit_JSON_{CITY_NAME}.html"

# Эти пути сохранены как параметры будущей ИИ-интеграции; сейчас модель не запускается.
MODEL_FILE = PROJECT_ROOT / "training" / "models" / "gguf" / "unsloth--Qwen3.5-4B-GGUF" / "Qwen3.5-4B-Q4_K_M.gguf"
LLAMA_SERVER = PROJECT_ROOT / "training" / "models" / "llama_cpp" / "cuda" / "llama-server.exe"
for label, path in (("JSON",REPORT_JSON),("digWater",DIGWATER),("Модель (не используется)",MODEL_FILE),("Сервер (не используется)",LLAMA_SERVER)):
    print(f"{label}: {path} — {'найден' if path.exists() else 'НЕ НАЙДЕН'}")
```

## Формирование HTML

Зависимость: `python -m pip install "ijson>=3.2.3"`. Для временной исправленной копии JSON требуется свободное место примерно по размеру файла.


```python
from reporting.json_report import render_json_report

if not REPORT_JSON.is_file():
    raise FileNotFoundError(REPORT_JSON)
if not DIGWATER.is_file():
    raise FileNotFoundError(DIGWATER)
result = render_json_report(REPORT_JSON, DIGWATER, OUTPUT_HTML)
print("Отчёт сохранён:", result["output_path"])
for group, count in result["counts"].items():
    if count:
        print(f"{group}: {count} объектов; сопоставлено с digWater: {result['matched'][group]}")
print("Откройте HTML-файл в браузере.")
```
